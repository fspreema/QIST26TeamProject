```python
# Import necessary libraries
import matplotlib.pyplot as plt
from RandomBondIsing import RandomBondIsing
```

***<u>Test Current Model Implementation:<u>***

```python
# Initialize the model parameters 
# -> Here we also define out geometry type and its corresponding attributes
model_params = {
    "lattice": "Square",
    "Lx": 10,
    "Ly": 10,
    "p": 0.1,
    "bc_x": "periodic",
    "bc_y": "periodic",
}

# Create an instance of the RandomBondIsing model
model = RandomBondIsing(model_params)
```

```python
# Print out all Coupling Terms
print(model.all_coupling_terms().to_TermList())
```

```python
# Plot Model and ordering
plt.figure(figsize=(5, 6))
ax = plt.gca()
model.lat.plot_coupling(ax)
model.lat.plot_sites(ax)
model.lat.plot_order(ax)
plt.show()
```

```python
# We first construct for the current model the matrix exponential of the hamiltonian by the following method
beta = 1
mtx_exp = model.H_MPO.make_U(-beta, "II")

print(mtx_exp)
```
